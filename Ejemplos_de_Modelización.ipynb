## Ejemplo 2
Una petroquímica se dedica a producir plástico para botellas (PET botella) y fibra textil (PET fibra) a partir de PTA (ácido tereftálico puro) y de EtG (Etilenglicol). Las cantidades de cada producto (en Tm) necesarias para producir una Tm de cada uno de estos derivados vienen dadas por:

| · | PET-Botella | PET-Fibra |
|---|---|---|
| PTA | 0.966 | 0.912 |
| EtG | 0.365 | 0.344 |

La venta en el mercado de PET-Botella da un beneficio de 60€ por Tm, mientras que la venta de PET-Fibra da un beneficio de 50€ por Tm.

Sean:
* $x_1$: Tm de PET-Botella producidas.
* $x_2$: Tm de PET-Fibra producidas.

**Función objetivo:**
$$ \text{Maximizar } Z = 60x_1 + 50x_2 $$

**Restricciones:**
* Disponibilidad de PTA: $$ 0.966x_1 + 0.912x_2 \leq 260000 $$
* Disponibilidad de EtG: $$ 0.365x_1 + 0.344x_2 \leq 150000 $$

# Ejemplo 3
Una empresa avícola tiene que elegir la dieta de sus animales. Para cumplir con la normativa sobre alimentación de aves para consumo humano las aves deben ingerir diariamente al menos
- 6 unidades del nutriente N1,
- 9 unidades del nutriente N2 y
- 7 unidades del nutriente N3.

Para ello dispone de tres piensos diferentes cuya composición de los tres nutrientes es la siguiente:

* un kilo del pienso P1 contiene 1 unidad de N1, 3 del N2 y 2 del N3;

* un kilo del pienso P2 contiene 2 unidades de N1, 1 de N2 y 1 del N3;

* y un kilo del pienso P3 contiene 3 unidades de N1, 2 de N2 y 3 de N3.

Si el coste de un kilo de pienso es de 6 u.m., 4 u.m. y 9 u.m. para P1, P2 y P3, respectivamente. Formula el modelo matemático que satisfaciendo
los requisitos mínimos nutricionales permita encontrar la dieta más
económica para la empresa.

**Modelo matemático de Programación Lineal:**

Variables:
- $x_1$ = kg de $P_1$
- $x_2$ = kg de $P_2$
- $x_3$ = kg de $P_3$

Función Objetivo
### $6x_1 + 4 x_2 + 9 x_3$

$Minimizar$

Restricciones:

- $1·x_1 + 2·x_2 + 3·x_3 ≥ 6$ (N1)
- $3·x_1 + 1·x_2 + 2·x_3 ≥ 9$ (N2)
- $2·x_1 + 1·x_2 + 3·x_3 ≥ 7$ (N3)

* No negatividad:
$$ x_1, x_2, x_3 \geq 0 $$


$$ x_1, x_2, x_3 ∈ ℝ^+ $$

# Ejemplo 4

La gerencia de una planta termoeléctrica de producción de electricidad, que emplea carbón como combustible, está estudiando la configuración operativa de la planta con el fin de cumplir nuevas leyes de contaminación medioambiental.

Las nuevas leyes especifican las siguientes tasas máximas de emisión:

* Óxido de azufre: 3000 partes por millón (PPM)
* Partículas (humos): 12 Kg./hora

El carbón se transporta con una cinta transportadora desde los silos a un pulverizador. La cinta transportadora puede manejar a lo sumo 20 ton/h de carbón, independientemente del tipo. El carbón se pulveriza y se inyecta en forma de polvo a una cámara de combustión. El calor producido se utiliza para generar vapor que impulsa unas turbinas que producen electricidad.

Actualmente la empresa utiliza un tipo de carbón barato y blando, llamado "B", pero que es de quema muy sucia. Existe la posibilidad de utilizar un carbón de tipo "A", más caro y duro, pero de quema más limpia. El carbón B produce 20000 lb de vapor por tonelada, mientras que el A produce 24000 lb/ton.

Las emisiones para cada uno de los tipos de carbón es la siguiente:

| Carbón | Óxido de azufre (PPM) | Partículas (kg/Ton) |
| :---: | :---: | :---: |
| A | 1800 | 0.5 |
| B | 3800 | 1.0 |

La pulverizadora solo puede pulverizar, por hora, a lo sumo 16 ton de carbón A ó 24 ton de carbón B, pudiéndose también pulverizar una mezcla de ambos.

Gerencia quiere contestar a una simple pregunta:
¿Cuál es la máxima producción posible de electricidad de la planta?

La respuesta debe ajustarse a los límites de emisión, las características de la planta y los tipos disponibles de carbón.

La respuesta permitirá determinar el margen de seguridad disponible para cubrir las demandas puntas de energía.

# Ejemplo 6

Una empresa tiene que contratar a 3 trabajadores para 3 puestos de trabajo (T1, T2, T3). Ha valorado a 5 candidatos que han sido valorados en una escala de 0 a 10 su habilidad para cada puesto de la siguiente forma:

| Puesto \ Candidato | 1 | 2 | 3 | 4 | 5 |
| :---: | :---: | :---: | :---: | :---: | :---: |
| **T1** | 3 | 5 | 7 | 6 | 6 |
| **T2** | 7 | 6 | 4 | 5 | 7 |
| **T3** | 6 | 8 | 6 | 9 | 4 |

* Formula el modelo matemático que maximiza la suma de las habilidades de los contratados.
* Formula el modelo matemático que maximiza el producto de las habilidades de los contratados.
* Formula el modelo matemático que maximiza el mínimo de las habilidades de los contratados.

### Modelo Binario: Max Suma

**Variables:**
- $b_{ij} \in \{0, 1\}$ := 1 si se contrata al candidato $i$ para el puesto $j$, 0 en caso contrario.
- $i \in \{1, 2, 3, 4, 5\}$ (Candidatos)
- $j \in \{1, 2, 3\}$ (Puestos T1, T2, T3)

**Función Objetivo:**
Maximizar la suma de las habilidades de los contratados:

$$
\begin{aligned}
\text{Max } Z = & \quad 3b_{11} + 5b_{21} + 7b_{31} + 6b_{41} + 6b_{51} \\
                & + 7b_{12} + 6b_{22} + 4b_{32} + 5b_{42} + 7b_{52} \\
                & + 6b_{13} + 8b_{23} + 6b_{33} + 9b_{43} + 4b_{53}
\end{aligned}
$$

**Restricciones:**

1. **Cada puesto debe ser cubierto por exactamente un candidato:**
   - $b_{11} + b_{21} + b_{31} + b_{41} + b_{51} = 1$ (Puesto T1)
   - $b_{12} + b_{22} + b_{32} + b_{42} + b_{52} = 1$ (Puesto T2)
   - $b_{13} + b_{23} + b_{33} + b_{43} + b_{53} = 1$ (Puesto T3)

2. **Cada candidato solo puede ser contratado para un puesto como máximo:**
   - $b_{11} + b_{12} + b_{13} \leq 1$ (Candidato 1)
   - $b_{21} + b_{22} + b_{23} \leq 1$ (Candidato 2)
   - $b_{31} + b_{32} + b_{33} \leq 1$ (Candidato 3)
   - $b_{41} + b_{42} + b_{43} \leq 1$ (Candidato 4)
   - $b_{51} + b_{52} + b_{53} \leq 1$ (Candidato 5)

### Modelo Binario: Max Product

**Variables:**
- $b_{ij} \in \{0, 1\}$ := 1 si se contrata al candidato $i$ para el puesto $j$, 0 en caso contrario.
- $i \in \{1, 2, 3, 4, 5\}$ (Candidatos)
- $j \in \{1, 2, 3\}$ (Puestos T1, T2, T3)

**Función Objetivo (No Lineal):**
Maximizar el producto de las habilidades de los contratados. Como solo se selecciona un candidato por puesto, podemos multiplicar la suma de habilidades de cada puesto:

$$
\begin{aligned}
\text{Max } Z = & (3b_{11} + 5b_{21} + 7b_{31} + 6b_{41} + 6b_{51}) \\
                \times & (7b_{12} + 6b_{22} + 4b_{32} + 5b_{42} + 7b_{52}) \\
                \times & (6b_{13} + 8b_{23} + 6b_{33} + 9b_{43} + 4b_{53})
\end{aligned}
$$

*Nota: Este modelo se puede linealizar maximizando la suma de los logaritmos de las habilidades (ya que maximizar el producto equivale a maximizar la suma de sus logaritmos):*
$$ \text{Max } Z' = \sum_{j=1}^{3} \sum_{i=1}^{5} \ln(h_{ij}) \cdot b_{ij} $$

**Restricciones:**

1. **Cada puesto debe ser cubierto por exactamente un candidato:**
   - $b_{11} + b_{21} + b_{31} + b_{41} + b_{51} = 1$ (Puesto T1)
   - $b_{12} + b_{22} + b_{32} + b_{42} + b_{52} = 1$ (Puesto T2)
   - $b_{13} + b_{23} + b_{33} + b_{43} + b_{53} = 1$ (Puesto T3)

2. **Cada candidato solo puede ser contratado para un puesto como máximo:**
   - $b_{11} + b_{12} + b_{13} \leq 1$ (Candidato 1)
   - $b_{21} + b_{22} + b_{23} \leq 1$ (Candidato 2)
   - $b_{31} + b_{32} + b_{33} \leq 1$ (Candidato 3)
   - $b_{41} + b_{42} + b_{43} \leq 1$ (Candidato 4)
   - $b_{51} + b_{52} + b_{53} \leq 1$ (Candidato 5)

### Modelo Binario: Max Min (Maximizar el mínimo)

Para linealizar la función objetivo de "maximizar el mínimo", introducimos una variable auxiliar $W$ (continua o entera) que representará el valor mínimo de las habilidades de los 3 candidatos contratados. Nuestro objetivo será maximizar $W$.

**Variables:**
- $b_{ij} \in \{0, 1\}$ := 1 si se contrata al candidato $i$ para el puesto $j$, 0 en caso contrario.
- $i \in \{1, 2, 3, 4, 5\}$ (Candidatos)
- $j \in \{1, 2, 3\}$ (Puestos T1, T2, T3)
- $W \geq 0$ := Nivel de habilidad mínimo entre los contratados.

**Función Objetivo:**
Maximizar la habilidad mínima garantizada:
$$ \text{Max } Z = W $$

**Restricciones:**

1. **$W$ debe ser menor o igual a la habilidad de la persona contratada para cada puesto:**
   - $W \leq 3b_{11} + 5b_{21} + 7b_{31} + 6b_{41} + 6b_{51}$ (Límite para el Puesto T1)
   - $W \leq 7b_{12} + 6b_{22} + 4b_{32} + 5b_{42} + 7b_{52}$ (Límite para el Puesto T2)
   - $W \leq 6b_{13} + 8b_{23} + 6b_{33} + 9b_{43} + 4b_{53}$ (Límite para el Puesto T3)

2. **Cada puesto debe ser cubierto por exactamente un candidato:**
   - $b_{11} + b_{21} + b_{31} + b_{41} + b_{51} = 1$ (Puesto T1)
   - $b_{12} + b_{22} + b_{32} + b_{42} + b_{52} = 1$ (Puesto T2)
   - $b_{13} + b_{23} + b_{33} + b_{43} + b_{53} = 1$ (Puesto T3)

3. **Cada candidato solo puede ser contratado para un puesto como máximo:**
   - $b_{11} + b_{12} + b_{13} \leq 1$ (Candidato 1)
   - $b_{21} + b_{22} + b_{23} \leq 1$ (Candidato 2)
   - $b_{31} + b_{32} + b_{33} \leq 1$ (Candidato 3)
   - $b_{41} + b_{42} + b_{43} \leq 1$ (Candidato 4)
   - $b_{51} + b_{52} + b_{53} \leq 1$ (Candidato 5)

### Modelo Binario: Max Min (Alternativa con variables de contratación y Big-M)

Esta es una formulación alternativa muy útil, especialmente cuando no sabemos de antemano cuántos puestos se van a cubrir o a cuántos candidatos vamos a contratar en total. Introducimos una variable que nos indica explícitamente si un candidato fue contratado o no.

**Variables:**
- $b_{ij} \in \{0, 1\}$ := 1 si se contrata al candidato $i$ para el puesto $j$, 0 en caso contrario.
- $y_i \in \{0, 1\}$ := 1 si el candidato $i$ es contratado (para cualquier puesto), 0 en caso contrario.
- $W \geq 0$ := Nivel de habilidad mínimo garantizado.
- $M$ := Un número lo suficientemente grande (Big-M), por ejemplo $M=10$ (ya que la habilidad máxima es 10).

**Función Objetivo:**
$$ \text{Max } Z = W $$

**Restricciones:**

1. **Cada puesto debe ser cubierto por exactamente un candidato:**
   - $b_{11} + b_{21} + b_{31} + b_{41} + b_{51} = 1$ (Puesto T1)
   - $b_{12} + b_{22} + b_{32} + b_{42} + b_{52} = 1$ (Puesto T2)
   - $b_{13} + b_{23} + b_{33} + b_{43} + b_{53} = 1$ (Puesto T3)

2. **Relación entre variables de asignación ($b_{ij}$) y contratación ($y_i$):**
   *(Si un candidato es asignado a algún puesto, entonces $y_i=1$. Además, esto asegura que un candidato tome máximo un puesto)*
   - $b_{11} + b_{12} + b_{13} = y_1$ (Candidato 1)
   - $b_{21} + b_{22} + b_{23} = y_2$ (Candidato 2)
   - $b_{31} + b_{32} + b_{33} = y_3$ (Candidato 3)
   - $b_{41} + b_{42} + b_{43} = y_4$ (Candidato 4)
   - $b_{51} + b_{52} + b_{53} = y_5$ (Candidato 5)

3. **Cálculo del mínimo solo para los candidatos contratados (usando Big-M):**
   *(Si el candidato no es contratado, $y_i=0$, la restricción queda $W \leq 0 + M$, desactivándose. Si es contratado, $y_i=1$, queda $W \leq \text{habilidad adquirida}$)*
   - $W \leq 3b_{11} + 7b_{12} + 6b_{13} + M(1 - y_1)$ (Candidato 1)
   - $W \leq 5b_{21} + 6b_{22} + 8b_{23} + M(1 - y_2)$ (Candidato 2)
   - $W \leq 7b_{31} + 4b_{32} + 6b_{33} + M(1 - y_3)$ (Candidato 3)
   - $W \leq 6b_{41} + 5b_{42} + 9b_{43} + M(1 - y_4)$ (Candidato 4)
   - $W \leq 6b_{51} + 7b_{52} + 4b_{53} + M(1 - y_5)$ (Candidato 5)
